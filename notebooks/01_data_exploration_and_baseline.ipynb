```python
def has_duplication(text, chunk_size=100, chunk_start=500):
    if len(text) < chunk_start + chunk_size:
        return False  # resume too short to even check this way
    chunk = text[chunk_start:chunk_start + chunk_size]
    return text.count(chunk) > 1

df_resumes_clean['has_duplication'] = df_resumes_clean['Resume_str'].apply(has_duplication)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

sample_docs = [
    "the cat sat on the mat",
    "the dog sat on the log",
    "cats and dogs are great pets"
]

vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(sample_docs)

print(tfidf_matrix.shape)
```

```python
print(vectorizer.get_feature_names_out())
```

```python
import pandas as pd

tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=vectorizer.get_feature_names_out()
)
tfidf_df
```

```python
# Same 2 sentences as before, but now with 10 extra sentences that all contain "the"
sample_docs_v2 = [
    "the cat sat on the mat",
    "the dog sat on the log",
    "cats and dogs are great pets",
    "the sun is bright today",
    "the weather is nice outside",
    "the man walked to the store",
    "the woman read the book",
    "the children played in the park",
    "the car drove down the road",
    "the bird flew over the tree",
    "the fish swam in the lake",
    "the sky turned dark tonight",
]

vectorizer_v2 = TfidfVectorizer()
tfidf_matrix_v2 = vectorizer_v2.fit_transform(sample_docs_v2)

tfidf_df_v2 = pd.DataFrame(
    tfidf_matrix_v2.toarray(),
    columns=vectorizer_v2.get_feature_names_out()
)

print("'the' score in sentence 0, WITH only 3 docs:", tfidf_df.loc[0, 'the'])
print("'the' score in sentence 0, WITH 12 docs:", tfidf_df_v2.loc[0, 'the'])
```

```python
all_texts = list(df_resumes_clean['Resume_str']) + list(df_jobdesc_clean['description'])

vectorizer_full = TfidfVectorizer()
tfidf_full = vectorizer_full.fit_transform(all_texts)

print(tfidf_full.shape)
```

```python
resume_0_vector = tfidf_full[0]
job_0_vector = tfidf_full[2423]
job_50_vector = tfidf_full[2423 + 50]

correct_match_score = cosine_similarity(resume_0_vector, job_0_vector)
mismatch_score = cosine_similarity(resume_0_vector, job_50_vector)

print("Resume 0 vs Job 0 (original pair):", correct_match_score)
print("Resume 0 vs Job 50 (comparison pair):", mismatch_score)
```

```python
!pip install -U sentence-transformers -q

from sentence_transformers import SentenceTransformer

model = SentenceTransformer('all-MiniLM-L6-v2')
print("Model loaded successfully")
```

```python
sentences = [
    "I am a software developer",
    "I am a programmer",
    "I love eating pizza"
]

embeddings = model.encode(sentences)
print(embeddings.shape)
```

```python
similarity_dev_prog = cosine_similarity([embeddings[0]], [embeddings[1]])
similarity_dev_pizza = cosine_similarity([embeddings[0]], [embeddings[2]])

print("Developer vs Programmer:", similarity_dev_prog)
print("Developer vs Pizza:", similarity_dev_pizza)
```

```python
resume_0_text = df_resumes_clean.iloc[0]['Resume_str']
job_0_text = df_jobdesc_clean.iloc[0]['description']
job_50_text = df_jobdesc_clean.iloc[50]['description']

embeddings_real = model.encode([resume_0_text, job_0_text, job_50_text])

match_score_embed = cosine_similarity([embeddings_real[0]], [embeddings_real[1]])
mismatch_score_embed = cosine_similarity([embeddings_real[0]], [embeddings_real[2]])

print("Resume 0 vs Job 0 (embeddings):", match_score_embed)
print("Resume 0 vs Job 50 (embeddings):", mismatch_score_embed)
```
